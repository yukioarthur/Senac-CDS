```python
import pygwalker as pyg
import time
import pandas as pd
```

# Data loading

```python
url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2022-01.parquet"

df_pandas = pd.read_parquet(url)
```

```python
df_pandas = df_pandas.sample(20000)
```

```python
df_pandas["hour"] = df_pandas["tpep_pickup_datetime"].dt.hour
df_pandas["weekday"] = df_pandas["tpep_pickup_datetime"].dt.day_name()
```

# Iniciando o Pygwalker

```python
pyg.walk(df_pandas)
```
