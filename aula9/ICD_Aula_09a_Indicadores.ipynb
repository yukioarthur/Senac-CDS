# Indicador: Apenas número e variação

https://plotly.github.io/plotly.py-docs/generated/plotly.graph_objects.Indicator.html

```python
fig = go.Figure(go.Indicator(
    mode="number+delta",
    value=media,
    delta={
        "reference":4,
        "relative":True
    },
    title={"text":"Avaliação geral (média)"}
))

fig.show()
```

# Angular gauge

https://plotly.com/python/gauge-charts/

```python
fig = go.Figure(go.Indicator(
    mode="number+delta+gauge",
    value=media,
    delta={
        "reference":4,
        "relative":True
    },
    gauge={"axis":{"range":[1,5]},
           "shape": "angular",
           "threshold":{
            "line":{"color":"red","width":3},
            "value":4
        }},
    title={"text":"Avaliação geral (média)"}
))

fig.show()
```

# Bullet Gauge equivalente

https://plotly.com/python/indicator/#bullet-gauge

```python
fig = go.Figure(go.Indicator(
    mode="number+delta+gauge",
    value=media,
    delta={
        "reference":4,
        "relative":True
    },
    gauge={"axis":{"range":[1,5]},
           "shape": "bullet",
           "threshold":{
            "line":{"color":"red","width":3},
            "value":4
        }},
    title={"text":"Avaliação geral (média)"},
    domain={"x":[0.2,1],"y":[0,1]}
))

fig.show()
```

# Funnel

https://plotly.com/python/funnel-charts/

```python
counts = [
    (df['Avaliação Geral'] >= 1).sum(),
    (df['Avaliação Geral'] >= 2).sum(),
    (df['Avaliação Geral'] >= 3).sum(),
    (df['Avaliação Geral'] >= 4).sum(),
    (df['Avaliação Geral'] >= 5).sum()
]
```

```python
fig = go.Figure(go.Funnel(
    y=["≥ 1", "≥ 2", "≥ 3", "≥ 4", "≥ 5"],
    x=counts,
    textposition="inside",
    textinfo="value+percent initial",
    textfont={"color": "black", "size": 14},
    opacity=0.9,
    marker = {"color": ["deepskyblue", "lightsalmon", "tan", "teal", "silver"],"line": {"color": ["black", "black", "black", "black", "black"]}},
))

fig.show()
```

# Radar plot

```python
import plotly.graph_objects as go

v = df.select_dtypes("number").mean()

fig = go.Figure(go.Scatterpolar(
    r=v, theta=v.index, fill="toself"
))

fig.update_layout(
    polar_radialaxis=dict(visible=True, range=[1, 5])
)

fig.show()
```

# Série temporal

```python
serie = df.set_index("Data da Pesquisa").resample("D").size()
```

df.set_index("Data da Pesquisa")

Transforma a coluna "Data da Pesquisa" no índice do DataFrame.

Isso é necessário porque o resample() trabalha com um índice temporal.

.resample("D")

Agrupa os registros por dia.

"D" significa day.

.size()

Conta quantos registros existem em cada grupo.

```python
serie
```

```python
import plotly.graph_objects as go

fig = go.Figure(go.Scatter(
    x=serie.index,
    y=serie.values,
    mode="lines"
))

fig.update_layout(
    title="Pesquisas ao longo do tempo",
    xaxis_title="Data",
    yaxis_title="Número de pesquisas"
)

fig.show()
```

```python
serie = df.set_index("Data da Pesquisa").resample("ME").size()
```

```python
fig = go.Figure(go.Scatter(
    x=serie.index,
    y=serie.values,
    mode="lines"
))

fig.update_layout(
    title="Pesquisas por mês",
    xaxis_title="Mês",
    yaxis_title="Número de pesquisas"
)

fig.show()
```
